Praktikum 7.04 – Set dan Operasi Himpunan

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)
```

Praktikum 7.05 – Studi Kasus: Data Mahasiswa dan Mata Kuliah

```python
mhs1 = {"Algoritma", "Matematika", "Bahasa Inggris"}
mhs2 = {"Algoritma", "Basis Data", "Matematika"}

print("Mata kuliah Mhs1:", mhs1)
print("Mata kuliah Mhs2:", mhs2)
print("Mata kuliah yang sama:", mhs1 & mhs2)
```

Praktikum 7.06 – Studi Kasus Dictionary: Daftar Nilai Mahasiswa

```python
nilai_mahasiswa = {	"Andi": 80,
"Budi": 75,
"Cici": 90
}

for nama, nilai in nilai_mahasiswa.items():
  print(f"{nama}: {nilai}")
  rata = sum(nilai_mahasiswa.values()) / len(nilai_mahasiswa)

print(f"Rata-rata nilai: {rata:.2f}")
```

7.	Set dan Himpunan

```python
A = {"Matematika", "Algoritma", "Bahasa Inggris"}
B = {"Algoritma", "Basis Data", "Matematika"}

print("Set A:", A)
print("Set B:", B)
print("Union (A | B):", A | B)
print("Intersection (A & B):", A & B)
print("Difference (A - B):", A - B)
```

8.	Studi Kasus Dictionary

```python
nilai_mahasiswa = {
    "Andi": 85,
    "Budi": 90,
    "Cici": 78
}

print("Daftar Nilai Mahasiswa:")
for nama, nilai in nilai_mahasiswa.items():
  print(f"{nama}: {nilai}")

rata_rata = sum(nilai_mahasiswa.values()) / len(nilai_mahasiswa)

print(f"\nRata-rata nilai: {rata_rata:.2f}")
```

9.	Studi Kasus Set – Klub Mahasiswa

```python
klub_a = {"Andi", "Budi", "Cici"}
klub_b = {"Cici", "Deni", "Eka"}

anggota_seluruh = klub_a | klub_b
print("Daftar seluruh anggota (union):", anggota_seluruh)

anggota_kedua_klub = klub_a & klub_b
print("Daftar anggota di kedua klub (intersection):", anggota_kedua_klub)

anggota_hanya_di_a = klub_a - klub_b
print("Anggota yang hanya ada di Klub A:", anggota_hanya_di_a)
```
