6. String Formatting dengan F - String

```python
nama = input("Nama: ")
umur = input("Umur: ")
jurusan = input("Jurusan: ")

print(f"Nama: {nama}\nUmur: {umur}\nJurusan: {jurusan}")
```

7. Operasi Dalam F - String

```python
bil1 = float(input("Masukkan bilangan pertama: "))
bil2 = float(input("Masukkan bilangan kedua: "))

# Operasi
jumlah = bil1 + bil2
kali = bil1 * bil2
bagi = bil1 / bil2

print(f"Penjumlahan      : {jumlah}")
print(f"Perkalian        : {kali}")
print(f"Hasil pembagian  : {bagi:.3f}")
```

8. Casting Data

```python
angka_string = input("Masukkan angka (string): ")

angka_int = int(angka_string)
print(f"Kuadrat dari {angka_int}: {angka_int**2}")

angka_float = float(angka_string)
print(f"Bilangan float: {angka_float:.2f}")
```

9. Studi Kasus - Analisis Teks

```python
kalimat = input("Masukkan kalimat: ")

# Jumlah kata
kata_list = kalimat.split()
jumlah_kata = len(kata_list)
print(f"Jumlah kata: {jumlah_kata}")

# Kata terpanjang
kata_terpanjang = max(kata_list, key=len)
print(f"Kata terpanjang: {kata_terpanjang}")

# Daftar kata (urut abjad)
kata_sorted = sorted(kata_list, key=str.lower)
print("Daftar kata urut abjad:", kata_sorted)
```
