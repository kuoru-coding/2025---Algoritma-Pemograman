Praktikum 10.1 – Bubble Sort

```python
def bubble_sort(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j] > data[j+1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan:", angka)
print("Data setelah diurutkan", bubble_sort(angka))
```

Praktikum 10.2 – Selection Sort

```python
def selection_sort(data):
  n = len(data)
  for i in range(n):
    min_index = i
    for j in range(i + 1, n):
      if data[j] < data[min_index]:
        min_index = j
        data[i], data[min_index] = data[min_index], data[i]
        return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

Praktikum 10.3 – Insertion Sort

```python
def insertion_sort(data):
  n = len(data)
  for i in range(1,n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
      data[j + 1] = key
      return data

angka = [12, 11, 13, 5, 6]
print("Data sebelum diurutkan:", angka)
print("Data setelah Insertion Sort:", insertion_sort(angka))
```

Praktikum 10.4 – Studi Kasus: Mengurutkan Nilai Mahasiswa


```python
def bubble_sort_nilai(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j] [1] < data[j + 1] [1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print("Data sebelum diurutkan:", mahasiswa)
print("Data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

Praktikum 10.5 – Perbandingan Waktu Eksekusi Sorting

```python
import random, time

data = [random.randint(1, 1000) for _ in range(500)]

for func in [bubble_sort, selection_sort, insertion_sort]:
  salinan = data.copy()
  start = time.time()
  func(salinan)
  end = time.time()
  print(f"{func.__name__} membutuhkan waktu {end - start:.5f} detik")
```

Praktikum 10.6 – Sorting Data Barang (Nama & Harga)


```python
def bubble_sort_key(data, key_index=0, reverse=False):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
      else:
        if data[j][key_index] > data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
  return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]
print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), 1, reverse=True))
```

3.	Eksperimen Bubble Sort

```python
def bubble_sort_descending(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j] < data[j+1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

angka_descending = [56, 77, 11, 80, 22, 67, 90]
print("Data sebelum diurutkan (descending):", angka_descending)
print("Data setelah diurutkan (descending):", bubble_sort_descending(angka_descending.copy()))
```

4.	Eksperimen Selection Sort

```python
def selection_sort_ascending(data):
  n = len(data)
  for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n):
      if data[j] < data[min_index]:
        min_index = j
    data[i], data[min_index] = data[min_index], data[i]
  return data

angka_selection_asc = [56, 77, 11, 80, 22, 67, 90]
print("Data sebelum diurutkan :", angka_selection_asc)
print("Data setelah Selection Sort :", selection_sort_ascending(angka_selection_asc.copy()))
```

5.	Eksperimen Insertion Sort



```python
def insertion_sort_strings(data):
  n = len(data)
  for i in range(1, n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

nama_mahasiswa = ["Budi", "Andi", "Cici", "Doni", "Eka"]
print("Nama mahasiswa sebelum diurutkan:", nama_mahasiswa)
print("Nama mahasiswa setelah diurutkan secara alfabet:", insertion_sort_strings(nama_mahasiswa.copy()))

nama_mahasiswa_2 = ["Zahra", "Yoga", "Xavier", "Wina", "Victor"]
print("\nNama mahasiswa sebelum diurutkan:", nama_mahasiswa_2)
print("Nama mahasiswa setelah diurutkan secara alfabet:", insertion_sort_strings(nama_mahasiswa_2.copy()))
```

7. Analisis Kompleksitas

```python
import random

def run_experiment(sort_func, data_size):
  data = [random.randint(1, 1000) for _ in range(data_size)]
  _, comparisons = sort_func(data.copy())
  return comparisons

sizes = [10, 50, 100]
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

results = {algo_name: {} for algo_name in algorithms.keys()}

print("Comparison Counts for Different Sorting Algorithms and Input Sizes:")
print("------------------------------------------------------------------")

for size in sizes:
  print(f"\nInput Size: {size}")
  for algo_name, func in algorithms.items():
    comp_count = run_experiment(func, size)
    results[algo_name][size] = comp_count
    print(f"  {algo_name}: {comp_count} comparisons")

print("\nSummary Table:")
print("------------------------------------------------------------------")
print(f"{'Algorithm':<15} | {'Size 10':<10} | {'Size 50':<10} | {'Size 100':<10}")
print(f"{'':-<15}-+-{'-':-<10}-+-{'-':-<10}-+-{'-':-<10}")
for algo_name, data in results.items():
  print(f"{algo_name:<15} | {data.get(10, '-'):<10} | {data.get(50, '-'):<10} | {data.get(100, '-'):<10}")
```

```python
1def flexible_sort(data, key_index=0, reverse=False):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
      else:
        if data[j][key_index] > data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
  return data

print("Defined 'flexible_sort' function.")
```

```python
def add_items_interactively(item_list=None):
  if item_list is None:
    item_list = []

  while True:
    item_name = input("Enter item name (or 'q' to quit): ")
    if item_name.lower() == 'q':
      break
    try:
      item_price = float(input(f"Enter price for {item_name}: "))
      item_list.append((item_name, item_price))
      print(f"'{item_name}' with price {item_price} added.")
    except ValueError:
      print("Invalid price. Please enter a numeric value.")
  return item_list

print("Defined 'add_items_interactively' function.")
```

```python
def display_items(item_list):
  if not item_list:
    print("No items to display.")
    return

  print("\n--- Current Items ---")
  for i, item in enumerate(item_list):
    print(f"{i + 1}. Name: {item[0]}, Price: {item[1]:.2f}")
  print("---------------------")

print("Defined 'display_items' function.")
```
