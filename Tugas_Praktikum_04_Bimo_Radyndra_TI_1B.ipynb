Praktikum 4.01 – Perulangan For dengan Range

```python
for i in range(1, 5):
   print(i)
```

Praktikum 4.02 – Perulangan While

```python
n = int(input("Masukkan angka: "))
while n > 0:
      print(n)
      n -= 1
```

Praktikum 03 – Nested Loop (Segitiga Bintang)

```python
for i in range(1, 5):
  for j in range(i):
    print("*", end="")
  print()

```

Praktikum 4.04 – Break dan Continue

```python
for i in range(1, 10):
  print(i)
  if i == 5:
    print("Angka 5 ditemukan")
    break
```

```python
for i in range(1, 5):
    if i % 2 == 1:
        continue
    print(i)
```

Praktikum 4.05 – For pada Dictionary

```python
data = {"Nama": "Andi", "Umur": 20, "Prodi": "TRK"}
for k, v in data.items():
  	print(k, ":", v)

```

Praktikum 4.06 – For dengan Enumerate

```python
daftar = ["Andi", "Budi", "Cici"]
for i, nama in enumerate(daftar, start=1):
  	print(i, nama)
```

Praktikum 4.07 – Studi Kasus: Program Kasir Multi-Item

```python
total = 0
while True:
  barang = input("Masukkan nama barang: ")
  if barang == "":
    break
  harga = int(input("Masukkan harga barang: "))
  jumlah = int(input("Masukkan jumlah barang: "))
  total += harga * jumlah

print("\n===== Struk Belanja =====")
print(f"Total Belanja : Rp {total}")
```

4.	Hitung Faktorial

```python
# Menghitung Faktorial dengan while

n = int(input("Masukkan sebuah bilangan: "))
faktorial = 1
i = 1
while i <= n:
    faktorial *= i
    i += 1
print(f"Faktorial dari {n} adalah {faktorial}")
```

5.	Pola Segitiga Terbalik

```python
# Mencetak segitiga bintang terbalik dengan tinggi 5 baris

tinggisegitiga = 5
for i in range(tinggisegitiga, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

6.	Cari Bilangan Prima dengan Break

```python
# Bilangan prima dengan Break


bilangan = int(input("Masukkan sebuah bilangan: "))
if bilangan > 1:
    for i in range(2, bilangan):
        if bilangan % i == 0:
            print(f"{bilangan} bukan bilangan prima karena habis dibagi {i}.")
            break
    else:
        print(f"{bilangan} adalah bilangan prima.")
else:
    print(f"{bilangan} bukan bilangan prima (bilangan harus lebih besar dari 1).")
```

10.	Studi Kasus – Kasir Diskon

```python
# Program Kasir Sederhana dengan Diskon 10%

total = 0  # Menyimpan total belanja

print("=== Program Kasir Sederhana ===")
print("Ketik 'selesai' jika sudah selesai berbelanja.\n")

while True:
    nama_barang = input("Masukkan nama barang: ")

    if nama_barang.lower() == "selesai":
        break

    harga = int(input(f"Masukkan harga {nama_barang}: "))
    jumlah = int(input(f"Masukkan jumlah {nama_barang}: "))

    subtotal = harga * jumlah
    total += subtotal

    print(f"Subtotal {nama_barang}: Rp{subtotal:,}\n")

# Hitung diskon jika total ≥ 200.000
if total >= 200000:
    diskon = total * 0.10
else:
    diskon = 0

total_bayar = total - diskon

# Tampilkan struk belanja
print("\n=== STRUK BELANJA ===")
print(f"Total Belanja : Rp{total:,}")
print(f"Diskon (10%)  : Rp{diskon:,}")
print(f"Total Bayar   : Rp{total_bayar:,}")
print("=========================")
print("Terima kasih telah berbelanja!")
```
