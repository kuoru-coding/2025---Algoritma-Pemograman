Praktikum 2 **Menghitung Segitiga [Rumus Heron]**

```
#  Menghitung luas segitiga

import math

a = float(input("masukkan sisi a: "))
b = float(input("masukkan sisi b: "))
c = float(input("masukkan si3si c: "))

s = (a + b + c) / 2
luas = math.sqrt (s * (s - a)  * (s - b)  * (s - c))
print("Luas segitiga =", luas)
```

Menghitung Volume Balok


```
# menghitung volume balok

# Input
p = float(input("Masukkan panjang balok: "))
l = float(input("Masukkan lebar balok: "))
t = float(input("Masukkan tinggi balok: "))

# Proses
volume = p * l * t

# Output
print("Volume balok =", volume)
```

Mengonversi detik

```
# Program konversi detik ke jam:menit:detik

# Input
total_detik = int(input("Masukkan waktu dalam detik: "))

# Proses
jam = total_detik // 3600
sisa = total_detik % 3600
menit = sisa // 60
detik = sisa % 60

# Output
print("Hasil konversi:")
print(jam, "jam", menit, "menit", detik, "detik")
```

Menghitung Rata - Rata Nilai Kuliah

```
# Program rata-rata nilai tiga mata kuliah

# Input
nilai1 = float(input("Masukkan nilai mata kuliah 1: "))
nilai2 = float(input("Masukkan nilai mata kuliah 2: "))
nilai3 = float(input("Masukkan nilai mata kuliah 3: "))

# Proses
rata = (nilai1 + nilai2 + nilai3) / 3

# Output
print("Rata-rata nilai =", rata)

if rata >= 60:
    print("Lulus")
else:
    print("Tidak Lulus")
```

Menghitung Diskon Total bayar

```
# Program menghitung total bayar setelah diskon

# Input
total_belanja = float(input("Masukkan total belanja: "))

# Proses
if total_belanja >= 500000:
    diskon = 0.2 * total_belanja
elif total_belanja >= 250000:
    diskon = 0.1 * total_belanja
else:
    diskon = 0

total_bayar = total_belanja - diskon

# Output
print("Total Belanja : Rp", total_belanja)
print("Diskon        : Rp", diskon)
print("Total Bayar   : Rp", total_bayar)
```
