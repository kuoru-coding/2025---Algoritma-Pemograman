Praktikum 3.05 – Match-Case

```python
# input menu = 2

menu = int(input("Pilih menu (1-3): "))
match menu:
  case 1:
    print("Anda memilih Nasi Goreng")
  case 2:
    print("Anda memilih Mie Ayam")
  case 3:
    print("Anda memilih Soto Ayam")
  case _:
    print("Menu tidak tersedia")
```

Praktikum 3.06 – Studi Kasus: Program Kasir

```python
# input nama barang = buku , harga = 60000 , jumlah = 5

barang = input("Masukkan nama barang: ")
harga = float(input("Masukkan harga barang: "))
jumlah = int(input("Masukkan jumlah barang: "))

total = harga * jumlah

if total >= 500000:
  diskon = 0.2
elif total >= 250000:
	diskon = 0.1
else:
	diskon = 0

bayar = total - (total * diskon)

print("\n===== Struk Belanja =====")
print(f"Barang : {barang}")
print(f"Jumlah : {jumlah}")
print(f"Total : Rp {total}")
print(f"Diskon : {diskon*100}%")
print(f"Bayar : Rp {bayar}")
```

Penugasan 3 No. 3 - Hitung Tarif Parkir

```python
jam = int(input("Masukkan lama parkir (jam): "))

if jam <= 2:
    tarif = 5000
elif jam > 2:
    tarif = 5000 + (jam - 2) * 2000

if tarif > 20000:
    tarif = 20000

print(f"Total tarif parkir: Rp {tarif}")
```

Penugasan 3 No. 7 - Studi Kasus Nyata - Program Penentuan Diskon

```python
nama_barang = input("Masukkan nama barang: ")
harga = int(input("Masukkan harga barang: "))
jumlah = int(input("Masukkan jumlah barang: "))

total = jumlah * harga

if total >= 500000:
    diskon = 0.20
elif total >= 250000:
    diskon = 0.10
else:
    diskon = 0

total_bayar = total - (total * diskon)

print("\n===== STRUK BELANJA =====")
print(f"Nama Barang   : {nama_barang}")
print(f"Harga         : Rp {harga:,}")
print(f"Jumlah        : {jumlah}")
print(f"Total Belanja : Rp {total:,}")
print(f"Diskon        : {diskon*100}%")
print(f"Total Bayar   : Rp {int(total_bayar):,}")
print("==========================")\
```
