Praktikum 01 - Pengenalan Variabel & Tipe Data

```python
# Integer (int)
umur = 18
print ("Umur:", umur, "->", type (umur))

# Float (float)
tinggi = 173.65
print ("Tinggi:", tinggi, "->", type (tinggi))

# String (str)
nama = "Bimo"
print ("Nama:", nama, "->", type(nama))

# Boolean (bool)
is_mahasiswa = True
print("Status Mahasiswa:", is_mahasiswa, "->", type(is_mahasiswa))

# List (list)
nilai = [6, 7, 67]
print ("Nilai:", nilai, "->", type(nilai))

# Tuple (tuble)
koordinat = (6, 7)
print("koordinat:", koordinat, "->", type(koordinat))

# Dictionary (dict)
mahasiswa = {"nama": "Bimo", "umur": 18, "prodi": "TRK"}
print("Data Mahasiswa:", mahasiswa, "->", type(mahasiswa))

# Set (set)
hobi =  {"membaca", "olahraga", "musik"}
print("Hobi:", hobi, "->", type(hobi))
```

Praktikum 02 – Input dan Output Dasar

```python
# Input dan Output Dasar

# Input string
nama = input("Masukkan nama Anda: ")

# Input integer
umur = int(input("Masukkan umur Anda: "))

# Input float
tinggi = float(input("Masukkan tinggi badan (cm): "))

# Output standar
print("Halo,", nama, "umur Anda:", umur, "tahun, tinggi:", tinggi, "cm")

# Output dengan format()
print("Halo {0}, umur Anda {1} tahun, tinggi {2} cm".format(nama, umur, tinggi))

# Output dengan f-string
print(f"Halo {nama}, umur Anda {umur} tahun, tinggi {tinggi} cm")
```

Praktikum 03 – Operator Aritmatika

```python
# Operator Matematika

a = 6
b = 7

print("a = 6 dan b = 7")
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
print("Hasil program menunjukkan bahwa Python mampu melakukan perhitungan dasar dengan operator yang sederhana dan intuitif.")
```

Praktikum 04 – Operator Relasional

```python
# Operator Relasional
x = 6
y = 7

print('x = 6 dan y = 7')
print("x == y :", x == y)
print("x != y :", x != y)
print("x < y :", x < y)
print("x > y :", x > y)
print("x <= y :", x <= y)
print("x >= y :", x >= y)
print('Dalam kehidupan nyata, operator relasional bisa dipakai untuk mengecek apakah nilai ujian ≥ 60,')
print('atau apakah kehadiran mahasiswa memenuhi batas minimal.')
```

Praktikum 05 – Operator Logika dengan Studi Kasus Nyata

```python
# Operator logika dasar
a = True
b = False
print("a and b =", a and b)
print("a or b =", a or b)
print("not a =", not a)
```

```python
# Studi kasus nyata: kelulusan
nilai = int(input("Masukkan nilai ujian: "))
hadir = int(input("Masukkan jumlah kehadiran: "))
lulus_nilai = nilai >= 60
lulus_hadir = hadir >= 12    # dari 16 pertemuan

if lulus_nilai and lulus_hadir:
  print("Status: LULUS")
else:
  print("Status: TIDAK LULUS")
```

Praktikum 06 – Program Kasir Sederhana

```python
barang = input("Masukkan nama barang: ")
harga = float (input('Masukkan harga barang: '))
jumlah = int(input("Masukkan jumlah barang: "))
total = harga * jumlah

# Aturan diskon 08: if total >= 500000:
diskon = 0.2
if total >= 250000:
  diskon = 0.1

else:
  diskon = 0

potongan = total * diskon
bayar = total - potongan
print("\n===== Struk Belanja =====")
print(f"Barang : {barang}")
print(f"Jumlah : {jumlah}")
print(f"Total : Rp {total}")
print(f"Diskon : {diskon * 100}%")
print(f"Bayar : Rp {bayar}")

```

IDENTITAS MAHASISWA

```python
# Mengambil Input
namapanjang = input("Masukkan nama Panjang Mahasiswa: ")
namapendek = input("Masukkan nama Panggilan Mahasiswa: ")
umur = int(input("Masukkan Umur Mahasiswa: "))
tinggi = float(input("Masukkan Tinggi Mahasiswa (cm): "))
status_in = input("Apakah Mahasiswa aktif? (y/n): ")

# Konversi boolean
status = "Aktif" if status_in.lower() == "y" else "Tidak Aktif"

# Menampilkan output
print(f"Selamat Datang, {namapendek}")
print("\n================================")
print("     IDENTITAS MAHASISWA")
print("================================")
print(f"Nama Lengkap : {namapanjang}")
print(f"Umur         : {umur} tahun")
print(f"Tinggi Badan : {tinggi} cm")
print(f"Status       : {status}")
print("================================")
```

OPERASI MATEMATIKA

```python
# Mengambil Input
a = (int(input("Masukkan nilai a: ")))
b = (int(input("Masukkan nilai b: ")))

# Operasi dan output
print(a, "+", b, "=", a + b)
print(a, "-", b, "=", a - b)
print(a, "*", b, "=", a * b)
print(a, "/", b, "=", a / b)
print(a, "//", b, "=", a // b)
print(a, "%", b, "=", a % b)
print(a, "**", b, "=", a ** b)
```

KONVERSI NILAI

```python
# Mengambil Input (float)
nilai_float = float(input("Masukkan sebuah angka desimal (contoh: 75.836): "))

# Konversi float ke integer (menghilangkan desimal)
nilai_integer = int(nilai_float)

# Konversi float ke string (menjadi teks)
nilai_string = str(nilai_float)

# Pembulatan 2 digit
nilai_round = round(nilai_float, 2)

# Menampilkan semua hasil
print("\n---------- HASIL KONVERSI & PEMBULATAN ----------")
print(f"Nilai Asli (float)      : {nilai_float}")
print(f"Dikonversi ke Integer   : {nilai_integer}")
print(f"Dikonversi ke String    : '{nilai_string}'")
print(f"Dibulatkan (2 desimal)  : {nilai_round}")
print("-----------------------------------------------")
```
