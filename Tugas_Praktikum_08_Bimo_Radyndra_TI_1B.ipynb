Praktikum 8.04 – Fungsi dengan *args dan **kwargs

```python
def jumlahkan(*args):
    return sum(args)

def biodata(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print("Total", jumlahkan(1, 2, 3, 4, 5))
biodata(Nama="Budi", Umur=20, Prodi="TRK")
```

Praktikum 8.05 – Scope Variabel

```python
x = 10

def ubah():
  x = 5
  print("Nilai lokal:", x)

ubah()
print("Nilai global:", x)
```

Praktikum 8.07 – Fungsi Decorator

```python
def decorator_sapa(fungsi):
  def wrapper():
    print("Sebelum Fungsi Dijalankan...")
    fungsi()
    print("Setelah Fungsi Dijalankan...")
  return wrapper

@decorator_sapa
def sapa():
    print("Halo, Mahasiswa Polines!")

sapa()
```

Praktikum 8.08 – Studi Kasus: Menghitung Rata-rata Nilai Mahasiswa

```python
def hitung_rata(data):
  return sum(data) / len(data)

nilai = [80, 90, 85]
rata = hitung_rata(nilai)

print("Rata - Rata nilai:", rata)
```

6. Fungsi dengan *args

```python
def jumlahkan_semua(*angka):
    total = sum(angka)
    return total

print("Total 3 bilangan (1, 2, 3):", jumlahkan_semua(1, 2, 3))
print("Total 5 bilangan (2, 4, 6, 8, 10):", jumlahkan_semua(2, 4, 6, 8, 10))
print("Total 7 bilangan (1, 3, 5, 7, 9, 11, 13):", jumlahkan_semua(1, 3, 5, 7, 9, 11, 13))
```

7. Fungsi dengan **kwargs

```python
def info_barang(**data):
    print("Informasi Barang:")
    for key, value in data.items():
        print(f"{key}: {value}")

info_barang(Nama="Buku", Harga=15000, Stok=30)
```

10. Fungsi Decorator

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print("Memulai fungsi...")
        hasil = func(*args, **kwargs)
        print("Fungsi selesai dijalankan.")
        return hasil
    return wrapper

@logger
def halo():
    print("Halo, selamat datang!")

halo()
@logger

def info_barang(**data):
    for key, value in data.items():
        print(f"{key}: {value}")

info_barang(Nama="Buku", Harga=15000, Stok=30)
```

11.	Studi Kasus Nyata – Program Nilai

```python

def input_nilai():

    nilai_mahasiswa = [80, 75, 55, 90, 60]
    return nilai_mahasiswa


def hitung_rata(nilai):
    return sum(nilai) / len(nilai)


def status_kelulusan(rata):
    if rata >= 60:
        return "Lulus"
    else:
        return "Tidak lulus"

nilai_akhir = lambda tugas, ujian: (0.4 * tugas) + (0.6 * ujian)


daftar_nilai = input_nilai()
rata = hitung_rata(daftar_nilai)
print("Nilai mahasiswa:", daftar_nilai)
print("Rata-rata nilai:", rata)
print("Status kelulusan:", status_kelulusan(rata))

tugas = 80
ujian = 70
print("Nilai akhir (40% tugas, 60% ujian):", nilai_akhir(tugas, ujian))
```
