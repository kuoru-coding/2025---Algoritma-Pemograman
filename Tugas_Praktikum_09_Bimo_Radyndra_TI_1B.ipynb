6.	Modul Buatan Sendiri

```python
%%writefile konversi.py

def celcius_ke_fahrenheit(celsius):
    """
    Mengkonversi suhu dari Celsius ke Fahrenheit.
    """
    return (celsius * 9/5) + 32

def fahrenheit_ke_celcius(fahrenheit):
    """
    Mengkonversi suhu dari Fahrenheit ke Celsius.
    """
    return (fahrenheit - 32) * 5/9
```

```python
import konversi


celsius_input = 15
fahrenheit_output = konversi.celcius_ke_fahrenheit(celsius_input)
print(f"{celsius_input} Celsius = {fahrenheit_output} Fahrenheit")

fahrenheit_input = 45
celsius_output = konversi.fahrenheit_ke_celcius(fahrenheit_input)
print(f"{fahrenheit_input} Fahrenheit = {celsius_output} Celsius")
```

  8.	Studi Kasus Mini

```python
%%writefile laporan.py

from pathlib import Path

def generate_report(score, total):
    text = f"Skor: {score}/{total}\nPersentase: {score/total*100:.0f}%\n"
    if score == total:
        text += "Status: Nilai sempurna!\n"
    else:
        text += "Status: Belum sempurna, coba lagi.\n"

    print(text)
    Path(__file__).parent.joinpath("laporan.txt").write_text(text)
```

```python
import random
from laporan import generate_report

score = 0
total_soal = 5

print("Selamat datang Kuis sederhana! Kerjakan soal berikut (5 soal)")

for i in range(1, total_soal + 1):
    a = random.randint(1, 10)
    b = random.randint(1, 10)
    try:
        jawab = int(input(f"Soal {i}: {a} + {b} = "))
    except ValueError:
        jawab = None
    if jawab == a + b:
        score += 1
        print("Benar!\n")
    else:
        print(f"Salah. Jawaban: {a + b}\n")

generate_report(score, total_soal)
```
