Praktikum 6.04 – List 2D (Matriks)

```python
matriks = [ [1, 2, 3],
            [4, 5, 6],
            [7, 8, 9]
                       ]

print(matriks[0][0])
print(matriks[1][2])
print(matriks[2])
```

Praktikum 6.05 – Operasi pada List 2D

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7, 8, 9], [10, 11, 12]]
C = []

for i in range(len(A)):
  baris = []
  for j in range(len(A[0])):
    baris.append(A[i][j] + B[i][j])
  C.append(baris)

print(C)
```

Praktikum 6.06 – Studi Kasus: Pengelolaan Nilai Mahasiswa

```python
nilai = [
    [80, 90, 85],
    [70, 75, 80],
    [88, 92, 84]
]

for i, baris in enumerate(nilai, start= 1):
  rata = sum(baris) / len(baris)
  print(f"Nilai rata-rata mahasiswa ke-{i}: {rata:.2f}")
```

4.	Operasi Dasar List

5.	Slicing List

```python
angka = list(range(1, 11))

print(angka[:3])
print(angka[-4:])
print(angka[::2])
```

6.	List 2D – Matriks

```python
matriks = [
    [7, 2, 7],
    [7, 6, 7]
]

print(matriks[0][1])
print(matriks[1][2])
print(matriks[0])
```

7.	Operasi Matriks 2D

```python
A = [[6, 7], [67, 67]]
B = [[6, 7], [67, 67]]
C = []

for i in range(len(A)):
  baris = []
  for j in range(len(A[0])):
    baris.append(A[i][j] + B[i][j])
  C.append(baris)

print(C)
```

8.	Rata-rata Nilai Mahasiswa

```python
nilai = [[75, 80, 85], [60, 70, 65], [90, 95, 92]]

for i, baris in enumerate(nilai, start=1):
  rata = sum(baris) / len(baris)
  print(f"Nilai rata-rata mahasiswa ke-{i}: {rata:.2f}")
```

9.	Studi Kasus – Data Toko

```python
stok = [["Buku", 20], ["Pulpen", 35], ["Penghapus", 15]]

# Menampilkan daftar barang dengan jumlah stoknya
print("Daftar Stok Barang:")
for item in stok:
  nama_barang = item[0]
  jumlah_stok = item[1]
  print(f"Barang: {nama_barang}, Stok: {jumlah_stok}")

# Mencari barang dengan stok terbanyak
stok_terbanyak = stok[0][1]
nama_barang_terbanyak = stok[0][0]

for item in stok[1:]:
  nama_barang = item[0]
  jumlah_stok = item[1]
  if jumlah_stok > stok_terbanyak:
    stok_terbanyak = jumlah_stok
    nama_barang_terbanyak = nama_barang

print(f"\nBarang dengan stok terbanyak: {nama_barang_terbanyak} (Stok: {stok_terbanyak})")

nama_barang_dicari = input("\nMasukkan nama barang yang ingin ditampilkan: ")
ditemukan = False
for item in stok:
  if item[0].lower() == nama_barang_dicari.lower():
    print(f"Stok {item[0]}: {item[1]}")
    ditemukan = True
    break
if not ditemukan:
  print(f"Barang '{nama_barang_dicari}' tidak ditemukan.")
```
